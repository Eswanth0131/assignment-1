```python
import os

LECTURE_URLS = "data/lecture_urls.txt"
YOUTUBE_IDS = "data/youtube_ids.txt"
PDF_COUNT = 59

def print_simple(title, path):
    print(title)
    if not os.path.exists(path):
        print("(file not found)\n")
        return
    
    with open(path, "r") as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
        for line in lines:
            print(line)
    print(f"Total: {len(lines)}\n")

print_simple("Websites", LECTURE_URLS)
print_simple("Videos", YOUTUBE_IDS)

print("PDFs")
print(f"Total: {PDF_COUNT}")
```
